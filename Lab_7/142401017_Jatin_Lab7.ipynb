```python
DOWNLOADS = "/home/142401017/Downloads/"
stop_words_set = set()
with open(os.path.join(DOWNLOADS,"en.txt"),"r") as f:
    stopwords = f.readlines()
    
    for i in range(len(stopwords)):
        stopwords[i] = stopwords[i][:-1]
        stop_words_set.add(stopwords[i])
```

```python
def tokenise(text,stop_words_set):
    body = text.lower()      
    lower_case_and_split = body.lower().split()
    
    ans = []
    
    for word in lower_case_and_split:
        word.strip(".,!?()[]:;")
    
        if len(word) > 1 and word.isalpha() and word not in stop_words_set:
            ans.append(word)

    return ans
```

```python
import os
import codecs

DATA_ROOT = "/home/142401017/Downloads/20news-bydate" 
TRAIN_DIR = os.path.join(DATA_ROOT, "20news-bydate-train")
TEST_DIR = os.path.join(DATA_ROOT, "20news-bydate-test")
```

```python
train_corpus = []
for folder in sorted(os.listdir(TRAIN_DIR)):
    FOLDER_PATH = os.path.join(TRAIN_DIR, str(folder))
    if os.path.isdir(FOLDER_PATH):
        for file in sorted(os.listdir(FOLDER_PATH)):
            CURR_FILE_PATH = os.path.join(FOLDER_PATH, file)
            with codecs.open(CURR_FILE_PATH, "r", encoding="utf-8", errors="ignore") as F:
                text = F.read()
                tokens = tokenise(text, stop_words_set) 
                train_corpus.append((tokens, folder))

test_corpus = []
for folder in sorted(os.listdir(TEST_DIR)):
    FOLDER_PATH = os.path.join(TEST_DIR, str(folder))
    if os.path.isdir(FOLDER_PATH):
        for file in sorted(os.listdir(FOLDER_PATH)):
            CURR_FILE_PATH = os.path.join(FOLDER_PATH, file)
            with codecs.open(CURR_FILE_PATH, "r", encoding="utf-8", errors="ignore") as F:
                text = F.read()
                tokens = tokenise(text, stop_words_set)
                test_corpus.append((tokens, folder))
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

vocabulary = []
word_to_index = {}
doc_frequency = {}

for tokens, label in train_corpus:
    unique_words_in_doc = set(tokens)
    for word in unique_words_in_doc:
        if word not in doc_frequency:
            doc_frequency[word] = 0
        doc_frequency[word] += 1

idx = 0
for word in doc_frequency:
    if doc_frequency[word] > 2: 
        vocabulary.append(word)
        word_to_index[word] = idx
        idx += 1

num_docs_train = len(train_corpus)
idf_dict = {}
for word in vocabulary:
    idf_dict[word] = math.log((num_docs_train) / (doc_frequency[word] + 1)) + 1

print("Vocabulary size :", len(vocabulary))
print(vocabulary)
```

```python
def trans_tfidf(corpus, vocabulary, word_to_index, idf_dict):
    num_docs = len(corpus)
    vocab_size = len(vocabulary)
    
    X_matrix = np.zeros((num_docs, vocab_size))
    
    for doc_idx in range(num_docs):
        tokens, label = corpus[doc_idx]

        tf_counts = {}
        for word in tokens:
            if word in tf_counts:
                tf_counts[word] = tf_counts[word] + 1
                
            else:
                tf_counts[word] = 1
                
        for word in tf_counts:
            if word in word_to_index:
                col_idx = word_to_index[word]
                raw_tf = tf_counts[word]
                log_tf = 1 + math.log(raw_tf) 
                
                X_matrix[doc_idx][col_idx] = log_tf * idf_dict[word]
                
    return X_matrix
```

```python
print("Task 1: sci.med vs sci.space")

task1_train = []
y_train1 = []

for tokens, folder in train_corpus:
    if folder == "sci.med" or folder == "sci.space":
        task1_train.append((tokens, folder))
        y_train1.append(folder)


task1_test = []
y_test1 = []
for tokens, folder in test_corpus:
    if folder == "sci.med" or folder == "sci.space":
        task1_test.append((tokens, folder))
        y_test1.append(folder)

X_train1 = trans_tfidf(task1_train, vocabulary, word_to_index, idf_dict)
X_test1 = trans_tfidf(task1_test, vocabulary, word_to_index, idf_dict)

clf1 = LogisticRegression(max_iter=1000)
clf1.fit(X_train1, y_train1)
y_pred1 = clf1.predict(X_test1)

print(f"Precision: {precision_score(y_test1, y_pred1, average='macro')}")
print(f"Recall:    {recall_score(y_test1, y_pred1, average='macro')}")
print(f"F1 Score:  {f1_score(y_test1, y_pred1, average='macro')}")
print()
print(confusion_matrix(y_test1, y_pred1, labels=clf1.classes_))
```

```python
def run_naive_bayes(train_data, test_data, target_classes):
    vocab_size = len(vocabulary)
    doc_counts_per_class = {}
    total_docs = len(train_data)
    
    for tokens, actual_class in train_data:
        if actual_class not in doc_counts_per_class:
            doc_counts_per_class[actual_class] = 0
            
        doc_counts_per_class[actual_class] += 1
            
    priors = {}
    for class_name in target_classes:
        class_docs = doc_counts_per_class.get(class_name, 0)
        if class_docs > 0:
            priors[class_name] = class_docs / total_docs
        else:
            priors[class_name] = 0

    nb_class_word_counts = {}
    nb_class_total_words = {}
    for class_name in target_classes:
        nb_class_word_counts[class_name] = {}
        nb_class_total_words[class_name] = 0
    
    for tokens, actual_class in train_data:
        for word in tokens:
            if word in word_to_index:
                if word not in nb_class_word_counts[actual_class]:
                    nb_class_word_counts[actual_class][word] = 0
                nb_class_word_counts[actual_class][word] += 1
                nb_class_total_words[actual_class] += 1

    y_pred = []
    for tokens, folder in test_data:
        best_class = None
        max_log_posterior = -float('inf')
        
        for class_name in target_classes:
            if priors[class_name] > 0:
                log_posterior = math.log(priors[class_name])
                
            else:
                log_posterior = math.log(1e-9)
            
            for word in tokens:
                if word in word_to_index:
                    word_count = nb_class_word_counts[class_name].get(word, 0)
                    word_prob = (word_count + 1) / (nb_class_total_words[class_name] + vocab_size)
                    log_posterior += math.log(word_prob)
                    
            if log_posterior > max_log_posterior:
                max_log_posterior = log_posterior
                best_class = class_name
                
        y_pred.append(best_class)
        
    return y_pred
```
